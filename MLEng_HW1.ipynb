```python
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
import numpy as np
from sklearn.model_selection import train_test_split

# Develop data set of 7th-root integers from 1-100

X = np.linspace(1,100,100)
y = (X ** (1/7))

#define keras model
model = Sequential()

model.add(Dense(6,input_dim=1,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(1))

#compile the keras model
opt = optimizers.Adam(learning_rate=0.001)
mse = tf.keras.losses.MeanSquaredError(
    reduction=tf.keras.losses.Reduction.SUM)
model.compile(loss=mse, optimizer=opt)

# split data
X_train, X_test, y_train, y_test=train_test_split(
    X,y,
    test_size=0.40,
    train_size=0.60,
    random_state=123,
    shuffle=True)

#fit the keras model on the dataset (CPU)
model.fit(X_train,y_train,epochs=1000,batch_size=10, verbose=0)
model.summary()

#make class predictions with the model

predictions = model.predict(X_test)

#summarize the first 10 cases
for i in range(10):
    print('%s => %.2f (expected %.2f)' %(X_test[i].tolist(), predictions[i], y_test[i]) )

import matplotlib.pyplot as plt
number_grid = np.linspace(1, 100, 100).reshape(-1, 1)*1.0
print("number_grid shape:", number_grid.shape)
print("number_grid dtype:", number_grid.dtype)
print("model input shape:", model.input_shape)
print("model built:", model.built)
# use model to predict 7th root of integers between 1 and 100
new_predict = model(number_grid, verbose=0).numpy()
new_predict = new_predict.flatten()
plt.scatter(X,y, label='true')
plt.plot(number_grid.flatten(),new_predict, color='red', label='model')
plt.xlabel('number')
plt.ylabel('7th root')
plt.legend()
```

```python
from sklearn import datasets
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
from sklearn.model_selection import train_test_split
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error

iris = datasets.load_iris()
iris.data
X = iris.data[:, 0].reshape(-1, 1) # only selecting sepal length
iris.feature_names
y = iris.data[:, 2] # only selecting petal length
X.shape
y.shape
# split data
X_train, X_test, y_train, y_test=train_test_split(
    X,y,
    test_size=0.40,
    train_size=0.60,
    random_state=123,
    shuffle=True)
model = Sequential()

model.add(Dense(8,input_dim=1,activation='relu'))
model.add(Dense(8,activation='relu'))
model.add(Dense(8,activation='relu'))
model.add(Dense(1))

#compile the keras model
opt = optimizers.Adam(learning_rate=0.001)
mse = tf.keras.losses.MeanSquaredError(
    reduction=tf.keras.losses.Reduction.SUM)
model.compile(loss=mse, optimizer=opt)


#fit the keras model on the dataset (CPU)
model.fit(X_train,y_train,epochs=1000,batch_size=10, verbose=0)
model.summary()

#make class predictions with the model

predictions = model.predict(X_test)
#summarize the first 10 cases
for i in range(10):
    print('%s => %.2f (expected %.2f)' %(X_test[i].tolist(), predictions[i], y_test[i]) )

mse = mean_squared_error(y_test, predictions)
print("Mean Squared Error:", mse)
```
